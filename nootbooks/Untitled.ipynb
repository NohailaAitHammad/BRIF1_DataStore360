```python
import pandas as pd
```

```python
df_raw = pd.read_csv('../data/raw/Sample - Superstore.csv')
print(df_raw.shape)
print(df_raw.head())
```

```python
from sqlalchemy import create_engine
```

```python
engine = create_engine("postgresql+psycopg2://root:root@localhost/datastore360")
print('connected')
```

```python
with engine.connect() as connection:
    print("Connexion PostgreSQL réussie !")
```

```python
df_raw.to_sql("superstore_raw", engine, schema='staging', if_exists="append", index=False)
```
