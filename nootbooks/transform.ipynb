```python
df_clean['Délai Livraison'] = (df_clean['Ship Date'] - df_clean['Order Date']).dt.days
df_clean[['Délai Livraison', 'Ship Mode']].dtypes
dominant_ship_mode = (
    df_clean.dropna(subset=['Ship Mode'])
    .groupby('Délai Livraison')['Ship Mode']
    .agg(lambda x: x.mode()[0])
)
print(dominant_ship_mode)
```

```python
#Inverser les dates quand Ship < Order ET que l'inversion donne un délai plausible
#delai_apres_swaping = -df_clean.loc[df_clean['Délai Livraison']<0, 'Délai Livraison']
#condition_swap_valid = df_clean['Délai Livraison']<0 & (df_clean['Délai Livraison'].abs() <= 30)
#print("Lignes à inverser :", condition_swap_valid.sum())
#print("Lignes à supprimer :", (df_clean['Délai Livraison']<0 & ~condition_swap_valid).sum())
mask_swap = df_clean['Délai Livraison'] < 0

# Après inversion, le délai serait -Délai_actuel
delai_apres_swap = -df_clean.loc[mask_swap, 'Délai Livraison']
mask_swap_valide = mask_swap & (df_clean['Délai Livraison'].abs() <= 30)

print("Lignes à inverser :", mask_swap_valide.sum())
print("Lignes à supprimer :", (mask_swap & ~mask_swap_valide).sum())

# Inversion des dates
df_clean.loc[mask_swap_valide, ['Order Date', 'Ship Date']] = (
    df_clean.loc[mask_swap_valide, ['Ship Date', 'Order Date']].values
)

# 2b. Supprimer les lignes irrécupérables (|délai| > 30 jours)
df_clean = df_clean[df_clean['Délai Livraison'].abs() <= 30].copy()
```

*Prix Unitaire d'apres le prix total*

```python
mask_valide = (
    df_clean['Sales'].notna()
    & df_clean['Quantity'].notna() & (df_clean['Quantity'] > 0)
    & df_clean['Discount'].notna() & (df_clean['Discount'] < 1)
)
df_clean['Prix Unitaire'] = np.nan
df_clean.loc[mask_valide, 'Prix Unitaire'] = (
    df_clean.loc[mask_valide, 'Sales']
    / (df_clean.loc[mask_valide, 'Quantity'] * (1 - df_clean.loc[mask_valide, 'Discount']))
)
print("Prix Unitaire NaN :", df_clean['Prix Unitaire'].isna().sum())
print(df_clean['Prix Unitaire'].describe())
```

```python
prix_ref = (
    df_clean.dropna(subset=['Prix Unitaire'])       # lignes valides
    .groupby('Product Name')['Prix Unitaire']       # grouper par produit
    .agg(lambda x: x.mode()[0])                     # mode = valeur la plus fréquente
)

print(f"Produits dans le mapping : {len(prix_ref)}")
print(prix_ref.head())
```

```python
mask_imput = (
    df_clean['Sales'].isna()
    & df_clean['Quantity'].notna()
    & df_clean['Discount'].notna()
    & df_clean['Product Name'].notna()
)
# Récupérer le prix de référence par produit
prix_produit = df_clean.loc[mask_imput, 'Product Name'].map(prix_ref)
# Recalculer Sales
df_clean.loc[mask_imput, 'Sales'] = (
    df_clean.loc[mask_imput, 'Quantity']
    * prix_produit
    * (1 - df_clean.loc[mask_imput, 'Discount'])
)

print("Sales NaN restants :", df_clean['Sales'].isna().sum())
```

```python
# Combien de Sales on a imputés ?
print("Sales NaN au départ :", df_clean['Sales'].isna().sum())


# Vérifier la cohérence : recalculer Prix Unitaire et comparer
df_clean['Prix Unitaire'] = (
    df_clean['Sales']
    / (df_clean['Quantity'] * (1 - df_clean['Discount']))
)

df_clean['Ecart'] = (df_clean['Prix Unitaire'] - df_clean['Prix_Unitaire_Verif']).abs()
#print("Écart max :", df_clean['Ecart'].max())
print("Prix Unitaire NaN après imputation :", df_clean['Prix Unitaire'].isna().sum())
```

```python
import hashlib
import hmac

def pseudonymize_row(row, salt_key):
    # Concaténation de la valeur et du sel
    data = f"{row}{salt_key}".encode('utf-8')
    # Calcul du HMAC-SHA256
    hashed = hmac.new(salt_key.encode('utf-8'), data, hashlib.sha256).hexdigest()
    return hashed

df_clean['Customer Name'] = df_clean['Customer Name'].apply(lambda x: pseudonymize_row(x, 'SECRET_SALT_KEY'))
df_clean
```

```python
mask = df_clean['Sales'] > 0
df_clean.loc[mask, 'Marge'] = (
    df_clean.loc[mask, 'Profit'] / df_clean.loc[mask, 'Sales']
)
df_clean.loc[mask, 'Marge']
```

Marge = 0	Neutre (ni gain ni perte)
Marge < 0	Perte (vente à perte)
